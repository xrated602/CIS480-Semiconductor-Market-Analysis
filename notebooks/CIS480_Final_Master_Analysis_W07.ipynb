## Final Team ETL and Analytical Dataset

The following section consolidates the team's final ETL and KPI workflow into one standardized long-format dataset for Power BI and Python analysis.

The original A04 retrieval and first working vertical slice above are preserved. The final workflow below incorporates the finalized ETL structure and supporting measures developed by Rey while retaining the project's validated six-security scope:

- AMD
- NVIDIA (NVDA)
- Intel (INTC)
- Taiwan Semiconductor Manufacturing (TSM)
- Broadcom (AVGO)
- SPY benchmark

All analytical percentage fields remain numeric proportions in the primary dataset so they can be formatted correctly in Power BI.


```python
# Reshape the already-downloaded yfinance data into the final long-format schema

company_names = {
    "AMD": "AMD",
    "NVDA": "NVIDIA",
    "INTC": "Intel",
    "TSM": "TSMC",
    "AVGO": "Broadcom",
    "SPY": "SPY (S&P 500 Benchmark)"
}

processed_frames = []

for ticker in tickers:
    df_ticker = pd.DataFrame({
        "Trading_Date": data.index,
        "Adj_Close": data["Close"][ticker].values,
        "Volume": data["Volume"][ticker].values
    }).dropna(subset=["Adj_Close"]).reset_index(drop=True)

    # Primary return and risk fields
    df_ticker["Daily_Return"] = df_ticker["Adj_Close"].pct_change()

    start_price = df_ticker["Adj_Close"].iloc[0]
    df_ticker["Cumulative_Return"] = (
        df_ticker["Adj_Close"] / start_price
    ) - 1

    df_ticker["Rolling_Peak"] = df_ticker["Adj_Close"].cummax()
    df_ticker["Drawdown"] = (
        df_ticker["Adj_Close"] - df_ticker["Rolling_Peak"]
    ) / df_ticker["Rolling_Peak"]

    # Metadata
    df_ticker["Ticker"] = ticker
    df_ticker["Company"] = company_names[ticker]
    df_ticker["Asset_Class"] = (
        "Benchmark" if ticker == "SPY" else "Equity"
    )

    processed_frames.append(df_ticker)

clean_long_df = pd.concat(processed_frames, ignore_index=True)

clean_long_df["Trading_Date"] = pd.to_datetime(
    clean_long_df["Trading_Date"]
)

clean_long_df.sort_values(
    by=["Ticker", "Trading_Date"],
    inplace=True
)

clean_long_df.reset_index(drop=True, inplace=True)

print("Final long-format dataset shape:", clean_long_df.shape)
print("First trading date:", clean_long_df["Trading_Date"].min())
print("Last trading date:", clean_long_df["Trading_Date"].max())
print("Tickers:", sorted(clean_long_df["Ticker"].unique()))
```

## Supporting Measures

The final analytical dataset includes four supporting measures in addition to the three primary KPIs:

1. **Trading Day Count** — sequential trading-day index for each ticker.
2. **Normalized Price** — each security indexed to 100 on its first trading date.
3. **Benchmark Difference** — each security's daily return minus SPY's daily return.
4. **Annual Return** — return from the first available trading day to the last available trading day within each calendar year.


```python
# Trading-Day Count
clean_long_df["Trading_Day_Count"] = (
    clean_long_df.groupby("Ticker").cumcount() + 1
)

# Normalized Price: base value of 100 on the first trading date
p0_series = clean_long_df.groupby("Ticker")["Adj_Close"].transform("first")
clean_long_df["Normalized_Price"] = (
    clean_long_df["Adj_Close"] / p0_series
) * 100

# Benchmark Difference: ticker daily return minus SPY daily return
spy_returns = (
    clean_long_df[clean_long_df["Ticker"] == "SPY"]
    .set_index("Trading_Date")["Daily_Return"]
)

clean_long_df["SPY_Daily_Return"] = (
    clean_long_df["Trading_Date"].map(spy_returns)
)

clean_long_df["Benchmark_Difference"] = (
    clean_long_df["Daily_Return"]
    - clean_long_df["SPY_Daily_Return"]
)

# SPY compared with itself should be 0 where a daily return exists
clean_long_df.drop(columns=["SPY_Daily_Return"], inplace=True)

# Annual Return
clean_long_df["Year"] = clean_long_df["Trading_Date"].dt.year

clean_long_df["Annual_Return"] = (
    clean_long_df
    .groupby(["Ticker", "Year"])["Adj_Close"]
    .transform(lambda x: (x.iloc[-1] / x.iloc[0]) - 1)
)

# Final chronological ordering
clean_long_df.sort_values(
    by=["Trading_Date", "Ticker"],
    inplace=True
)

clean_long_df.reset_index(drop=True, inplace=True)

clean_long_df.head(12)
```

## Primary KPI Summary

The three primary project KPIs are:

- **Cumulative Return** — total price performance from the first to the last available trading date.
- **Annualized Volatility** — standard deviation of daily returns multiplied by the square root of 252 trading days.
- **Maximum Drawdown** — the largest decline from a running peak during the analysis period.


```python
kpi_summary = (
    clean_long_df
    .groupby("Ticker")
    .agg(
        Asset_Class=("Asset_Class", "first"),
        Start_Price=("Adj_Close", "first"),
        End_Price=("Adj_Close", "last"),
        Cumulative_Return=("Cumulative_Return", "last"),
        Annualized_Volatility=(
            "Daily_Return",
            lambda x: x.std() * np.sqrt(252)
        ),
        Max_Drawdown=("Drawdown", "min")
    )
    .reset_index()
)

kpi_summary
```

```python
# Display a presentation-friendly KPI table while keeping the source table numeric

kpi_summary_display = kpi_summary.copy()

for col in [
    "Cumulative_Return",
    "Annualized_Volatility",
    "Max_Drawdown"
]:
    kpi_summary_display[col] = (
        kpi_summary_display[col] * 100
    ).round(2).astype(str) + "%"

kpi_summary_display
```

## Week 6 Methodology Architecture and Baseline Validation

This section adds the Week 6 **Methodology Architecture and Baseline** evidence without changing the established Week 5 data-quality workflow.

### Smallest Complete Pipeline Path

The reproducible project path is:

**Yahoo Finance / yfinance → raw CSV snapshot → Python/Jupyter preprocessing → data-quality validation → KPI calculation → processed CSV → Power BI dashboard**

The Week 6 baseline test focuses on one inspectable calculation within that path: **SPY cumulative return** for the complete project observation period.

### Baseline and Evaluation Design

- **Project pathway:** Dashboard / analytical methodology
- **Evaluation unit:** SPY over the full project period (first through last available trading observation)
- **Sample input:** SPY first and last adjusted closing prices already present in the validated analytical dataset
- **Baseline formula:** `(Ending Adjusted Close / Beginning Adjusted Close) - 1`
- **Pipeline result:** SPY `Cumulative_Return` from the project's KPI summary
- **Primary measure:** Absolute difference between the independently recomputed baseline and pipeline result
- **Pass condition:** Difference must be no greater than **0.0001 as a proportion**, equivalent to **0.01 percentage point**
- **Error consequence:** A failed reconciliation would indicate that the displayed historical return could be inconsistent with the underlying validated prices and should not be accepted for dashboard use until investigated.
- **Leakage risk:** This project is descriptive rather than predictive. The test uses only observations inside the defined 2021–2025 project period. No future observation is used to predict an earlier observation.
- **Random cross-validation:** Not appropriate because the project does not train a predictive model. Calculation reconciliation and reproducible dashboard tests are better matched to the claim being made.

The test below derives the baseline directly from the validated `clean_long_df` prices and compares it with the independently produced KPI summary. It does not hard-code a PASS result.


```python
# ============================================================
# WEEK 6 REPRODUCIBLE BASELINE TEST
# SPY cumulative-return reconciliation
# ============================================================

test_ticker = "SPY"
tolerance = 0.0001  # proportion = 0.01 percentage point

# Sample input from the validated analytical dataset
spy_data = (
    clean_long_df.loc[clean_long_df["Ticker"] == test_ticker]
    .sort_values("Trading_Date")
    .copy()
)

baseline_start_price = float(spy_data["Adj_Close"].iloc[0])
baseline_end_price = float(spy_data["Adj_Close"].iloc[-1])

# Independent baseline calculation
baseline_cumulative_return = (
    baseline_end_price / baseline_start_price
) - 1

# Pipeline result produced by the project's KPI summary
pipeline_cumulative_return = float(
    kpi_summary.loc[
        kpi_summary["Ticker"] == test_ticker,
        "Cumulative_Return"
    ].iloc[0]
)

absolute_difference = abs(
    baseline_cumulative_return - pipeline_cumulative_return
)

week6_status = (
    "PASSED" if absolute_difference <= tolerance else "FAILED"
)

week6_validation_record = pd.DataFrame({
    "Test": ["SPY cumulative-return reconciliation"],
    "Input": [
        f"Start={baseline_start_price:.6f}; End={baseline_end_price:.6f}"
    ],
    "Expected_Result": [
        "Independent baseline agrees with pipeline within 0.01 percentage point"
    ],
    "Baseline_Result": [baseline_cumulative_return],
    "Pipeline_Result": [pipeline_cumulative_return],
    "Absolute_Difference": [absolute_difference],
    "Tolerance": [tolerance],
    "Status": [week6_status]
})

print("=" * 70)
print("CIS480 WEEK 6 BASELINE VALIDATION")
print("=" * 70)
print(f"Security: {test_ticker}")
print(f"Start adjusted close: {baseline_start_price:.6f}")
print(f"End adjusted close: {baseline_end_price:.6f}")
print(f"Baseline cumulative return: {baseline_cumulative_return:.8f} "
      f"({baseline_cumulative_return:.2%})")
print(f"Pipeline cumulative return: {pipeline_cumulative_return:.8f} "
      f"({pipeline_cumulative_return:.2%})")
print(f"Absolute difference: {absolute_difference:.10f}")
print(f"Tolerance: {tolerance:.4f} proportion (0.01 percentage point)")
print(f"WEEK 6 BASELINE STATUS: {week6_status}")
print("=" * 70)

assert absolute_difference <= tolerance, (
    "Week 6 baseline reconciliation failed: "
    f"difference {absolute_difference:.10f} exceeds tolerance {tolerance:.10f}"
)

week6_validation_record
```

```python
# Week 7 setup and evidence metadata
from datetime import datetime, timezone
from pathlib import Path

week7_run_utc = datetime.now(timezone.utc).isoformat(timespec="seconds")
week7_tickers = ["AMD", "NVDA", "INTC", "TSM", "AVGO", "SPY"]
week7_required = {"Ticker", "Trading_Date", "Adj_Close", "Daily_Return", "Drawdown"}
assert week7_required.issubset(clean_long_df.columns)
assert set(clean_long_df["Ticker"].unique()) == set(week7_tickers)
assert not clean_long_df.duplicated(["Ticker", "Trading_Date"]).any()
assert clean_long_df["Adj_Close"].notna().all()
assert (clean_long_df["Adj_Close"] > 0).all()
assert set(kpi_summary["Ticker"]) == set(week7_tickers)

week7_n_dates = clean_long_df.groupby("Ticker")["Trading_Date"].nunique()
week7_scope = pd.DataFrame({
    "Ticker": week7_n_dates.index,
    "Trading_Days": week7_n_dates.values,
    "First_Date": [clean_long_df.loc[clean_long_df.Ticker == t, "Trading_Date"].min().date().isoformat() for t in week7_n_dates.index],
    "Last_Date": [clean_long_df.loc[clean_long_df.Ticker == t, "Trading_Date"].max().date().isoformat() for t in week7_n_dates.index],
})
print("Week 7 executed at (UTC):", week7_run_utc)
print("Price basis: yfinance auto_adjust=True Close; daily return basis: pct_change")
display(week7_scope)
```

### Output 1 — Cumulative return and SPY comparison

**Unit:** numeric proportion in the CSV; percentage in the displayed table. **Denominator:** each security's first available adjusted closing price. **Baseline:** SPY's cumulative return over the same observation window. Difference is measured in **percentage points**, not percent change. A positive difference means historical outperformance of SPY over this period, not future superiority.

```python
spy_cumulative = float(kpi_summary.loc[kpi_summary.Ticker == "SPY", "Cumulative_Return"].iloc[0])
week7_returns = kpi_summary[["Ticker", "Start_Price", "End_Price", "Cumulative_Return"]].copy()
week7_returns["SPY_Cumulative_Return"] = spy_cumulative
week7_returns["Difference_vs_SPY"] = week7_returns["Cumulative_Return"] - spy_cumulative
week7_returns = week7_returns.sort_values("Cumulative_Return", ascending=False).reset_index(drop=True)
week7_returns.to_csv("week07_cumulative_return_vs_spy.csv", index=False)
display(week7_returns.style.format({"Start_Price":"{:,.4f}","End_Price":"{:,.4f}","Cumulative_Return":"{:.2%}","SPY_Cumulative_Return":"{:.2%}","Difference_vs_SPY":"{:+.2%}"}))
```

### Output 2 — Historical risk comparison

**Annualized volatility:** sample standard deviation of nonmissing daily returns multiplied by √252; denominator is the available daily return observations (sample SD uses *n−1*). **Maximum drawdown:** smallest `(price / cumulative historical peak) − 1` across available dates. **SPY comparison:** differences are in percentage points. Higher volatility and more negative maximum drawdown describe greater realized historical risk by these measures; they do not capture every risk dimension.

```python
spy_risk = kpi_summary.set_index("Ticker").loc["SPY"]
week7_risk = kpi_summary[["Ticker", "Annualized_Volatility", "Max_Drawdown"]].copy()
week7_risk["Volatility_Difference_vs_SPY"] = week7_risk["Annualized_Volatility"] - float(spy_risk["Annualized_Volatility"])
week7_risk["Drawdown_Difference_vs_SPY"] = week7_risk["Max_Drawdown"] - float(spy_risk["Max_Drawdown"])
week7_risk["Daily_Return_Observations"] = week7_risk["Ticker"].map(clean_long_df.groupby("Ticker")["Daily_Return"].count())
week7_risk = week7_risk.sort_values("Annualized_Volatility", ascending=False).reset_index(drop=True)
week7_risk.to_csv("week07_risk_vs_spy.csv", index=False)
display(week7_risk.style.format({c:"{:+.2%}" if "Difference" in c else "{:.2%}" for c in ["Annualized_Volatility", "Max_Drawdown", "Volatility_Difference_vs_SPY", "Drawdown_Difference_vs_SPY"]}))
```

### Independent validation — NVDA annualized volatility

**Question:** Does a calculation based directly on the NVDA adjusted closing-price sequence reconcile with the pipeline's annualized-volatility KPI? **Expected:** absolute difference ≤ 0.0001 proportion (0.01 percentage point). **Actual:** recorded by executable code below. Independent path uses explicit adjacent-price ratios and Python's `statistics.stdev` rather than the pipeline's pandas `Daily_Return.std()`. It verifies this KPI for one security, not all KPIs or the source data itself.

```python
import math
import statistics

nvda_prices = (clean_long_df.loc[clean_long_df.Ticker == "NVDA"]
               .sort_values("Trading_Date")["Adj_Close"].astype(float).tolist())
assert len(nvda_prices) >= 3
independent_daily_returns = [nvda_prices[i] / nvda_prices[i-1] - 1.0 for i in range(1, len(nvda_prices))]
independent_nvda_vol = statistics.stdev(independent_daily_returns) * math.sqrt(252)
pipeline_nvda_vol = float(kpi_summary.loc[kpi_summary.Ticker == "NVDA", "Annualized_Volatility"].iloc[0])
week7_tolerance = 0.0001
week7_abs_diff = abs(independent_nvda_vol - pipeline_nvda_vol)
week7_status = "PASS" if week7_abs_diff <= week7_tolerance else "FAIL"
week7_validation = pd.DataFrame([{
    "Test": "NVDA annualized volatility independent reconciliation",
    "Input_Question": "NVDA adjusted closes; daily simple returns; sample SD; 252-day scaling",
    "First_Date": str(week7_scope.set_index("Ticker").loc["NVDA", "First_Date"]),
    "Last_Date": str(week7_scope.set_index("Ticker").loc["NVDA", "Last_Date"]),
    "Daily_Return_Observations": len(independent_daily_returns),
    "Expected_Result": "Absolute difference <= 0.0001 proportion",
    "Actual_Independent": independent_nvda_vol,
    "Actual_Pipeline": pipeline_nvda_vol,
    "Absolute_Difference": week7_abs_diff,
    "Tolerance": week7_tolerance,
    "Status": week7_status,
    "Executed_UTC": week7_run_utc,
    "Evidence": "CIS480_Final_Master_Analysis_W07.ipynb / Week 7 independent validation cell"
}])
week7_validation.to_csv("week07_validation_record.csv", index=False)
display(week7_validation)
print("WEEK 7 VALIDATION STATUS:", week7_status)
assert week7_status == "PASS", "NVDA volatility reconciliation failed: investigate input and calculation before reporting results"
```

## Independent verification of NVDA annualized volatility

```python
import numpy as np
import pandas as pd

# --------------------------------------------------
# Configuration
# --------------------------------------------------
TICKER = "NVDA"
EXPECTED_OBSERVATIONS = 1254
EXPECTED_VOLATILITY_PCT = 52.2247
TOLERANCE = 0.0001  # Percentage points

# IMPORTANT:
# Set this to the notebook's actual UNROUNDED pipeline
# value expressed as a decimal (e.g., 0.522247...).
# Do not use the rounded 52.2247% display value.
notebook_volatility = kpi_summary.loc[kpi_summary['Ticker'] == 'NVDA', 'Annualized_Volatility'].item()

# --------------------------------------------------
# 1. Identify NVDA daily return values
# --------------------------------------------------
nvda = clean_long_df.loc[clean_long_df["Ticker"].eq(TICKER)].copy()

nvda["Trading_Date"] = pd.to_datetime(nvda["Trading_Date"])
nvda = nvda.sort_values("Trading_Date")

# Use the existing Daily_Return column that the notebook
# uses for its volatility calculation.
daily_returns = pd.to_numeric(
    nvda["Daily_Return"], errors="coerce"
)

# Exclude missing and infinite observations.
valid_returns = daily_returns[
    daily_returns.notna() & np.isfinite(daily_returns)
]

print("NVDA volatility validation")
print("=" * 55)

print("\n1. Daily return values identified")
print(f"PASS: {len(valid_returns)} valid values identified"
      if len(valid_returns) > 0 else
      "FAIL: No valid daily return values found")

print("\nFirst five valid daily returns:")
print(valid_returns.head().to_string())

# --------------------------------------------------
# 2. Confirm the number of valid observations
# --------------------------------------------------
observation_count = len(valid_returns)

count_pass = observation_count == EXPECTED_OBSERVATIONS

print("\n2. Valid observation count")
print(f"{'PASS' if count_pass else 'FAIL'}: "
      f"{observation_count} observations; "
      f"expected {EXPECTED_OBSERVATIONS}")

# --------------------------------------------------
# 3. Independently calculate SAMPLE standard deviation
# --------------------------------------------------
# pandas .std() defaults to ddof=1, the sample standard
# deviation. Explicitly set ddof=1 for clarity.
sample_std = valid_returns.std(ddof=1)

std_pass = (
    np.isfinite(sample_std)
    and observation_count == EXPECTED_OBSERVATIONS
)

print("\n3. Sample standard deviation")
print(f"{'PASS' if std_pass else 'FAIL'}: "
      f"{sample_std:.12f}")

# --------------------------------------------------
# 4. Annualize and convert to percentage
# --------------------------------------------------
annualized_volatility_decimal = sample_std * np.sqrt(252)
annualized_volatility_pct = annualized_volatility_decimal * 100

print("\n4. Independently calculated annualized volatility")
print(f"Annualized decimal: "
      f"{annualized_volatility_decimal:.12f}")
print(f"Annualized percent: "
      f"{annualized_volatility_pct:.10f}%")

# --------------------------------------------------
# 5. Compare with the notebook's unrounded value
# --------------------------------------------------
notebook_volatility_decimal = float(notebook_volatility)
notebook_volatility_pct = notebook_volatility_decimal * 100

# Both values are now expressed in percentage points.
absolute_difference_pct = abs(
    annualized_volatility_pct - notebook_volatility_pct
)

print("\n5. Notebook comparison")
print(f"Independent result: {annualized_volatility_pct:.10f}%")
print(f"Notebook result:    {notebook_volatility_pct:.10f}%")
print(f"Difference:         {absolute_difference_pct:.12f} "
      "percentage points")

# --------------------------------------------------
# 6. Apply the tolerance in consistent units
# --------------------------------------------------
tolerance_pass = (
    np.isfinite(absolute_difference_pct)
    and absolute_difference_pct <= TOLERANCE
)

print("\n6. Tolerance validation")
print(f"Tolerance: {TOLERANCE:.4f} percentage points")
print(f"{'PASS' if tolerance_pass else 'FAIL'}: "
      f"Absolute difference "
      f"{'within' if tolerance_pass else 'exceeds'} tolerance")

# --------------------------------------------------
# Final validation summary
# --------------------------------------------------
checks = {
    "1. NVDA daily returns identified": len(valid_returns) > 0,
    "2. Exactly 1,254 valid observations": count_pass,
    "3. Sample standard deviation calculated": std_pass,
    "4. Annualization and percent conversion valid": (
        np.isfinite(annualized_volatility_pct)
        and np.isfinite(annualized_volatility_decimal)
    ),
    "5. Notebook pipeline value available": (
        np.isfinite(notebook_volatility_decimal)
    ),
    "6. Absolute difference within tolerance": tolerance_pass,
}

# Checks 5 and 6 require a valid notebook reference value.
# The overall result also requires the expected observation count.
print("\n" + "=" * 55)
print("FINAL VALIDATION SUMMARY")

for description, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL'}: {description}")

overall_pass = all(checks.values())

print("=" * 55)
print(f"OVERALL RESULT: {'PASS' if overall_pass else 'FAIL'}")
```
