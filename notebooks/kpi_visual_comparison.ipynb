```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("kpi_summary_metrics_2021_2025.csv")
```

```python
# Convert decimal KPIs to percentages
df["Return_%"] = df["Cumulative_Return"] * 100
df["Volatility_%"] = df["Annualized_Volatility"] * 100
df["Drawdown_%"] = df["Max_Drawdown"] * 100
```

```python
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Cumulative return
sns.barplot(data=df, x="Ticker", y="Return_%", ax=axes[0])
axes[0].axhline(
    df.loc[df.Ticker == "SPY", "Return_%"].iloc[0],
    color="black", linestyle="--", linewidth=1.5
)
axes[0].set_title("Cumulative Return vs SPY")
axes[0].set_ylabel("Return (%)")
axes[0].set_xlabel("")

# Volatility
sns.barplot(data=df, x="Ticker", y="Volatility_%", ax=axes[1])
axes[1].axhline(
    df.loc[df.Ticker == "SPY", "Volatility_%"].iloc[0],
    color="black", linestyle="--", linewidth=1.5
)
axes[1].set_title("Annualized Volatility vs SPY")
axes[1].set_ylabel("Volatility (%)")
axes[1].set_xlabel("")

# Drawdown
sns.barplot(data=df, x="Ticker", y="Drawdown_%", ax=axes[2])
axes[2].axhline(
    df.loc[df.Ticker == "SPY", "Drawdown_%"].iloc[0],
    color="black", linestyle="--", linewidth=1.5
)
axes[2].set_title("Maximum Drawdown vs SPY")
axes[2].set_ylabel("Maximum Drawdown (%)")
axes[2].set_xlabel("")

plt.suptitle("Stock Performance vs SPY", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
```
