This section documents a small end-to-end path from representative historical-price input to a labeled analytical output. It validates the required data, calculates a defined return measure, and reports a comparison that can be checked before the full analysis continues.

```python
# Validate the downloaded input before analysis
required_price_field = "Close"
retrieved_tickers = sorted(data[required_price_field].columns.tolist())
missing_tickers = sorted(set(tickers) - set(retrieved_tickers))

print("Required price field present:", required_price_field in data.columns.get_level_values(0))
print("Requested tickers:", sorted(tickers))
print("Retrieved tickers:", retrieved_tickers)
print("Missing requested tickers:", missing_tickers)
print("First trading date:", data.index.min())
print("Last trading date:", data.index.max())
print("Duplicate dates:", int(data.index.duplicated().sum()))
print("Missing Close values by ticker:")
print(data[required_price_field].isna().sum())

validation_passed = (
    required_price_field in data.columns.get_level_values(0)
    and len(missing_tickers) == 0
    and data.index.duplicated().sum() == 0
)
print("\nA04 input validation status:", "PASSED" if validation_passed else "PARTIALLY PASSED / REVIEW NEEDED")
```

```python
# First-build calculation: cumulative return from first to last available Close
first_build = pd.DataFrame({
    "First_Close": data["Close"].apply(lambda s: s.dropna().iloc[0]),
    "Last_Close": data["Close"].apply(lambda s: s.dropna().iloc[-1])
})
first_build["Cumulative_Return_Pct"] = (
    (first_build["Last_Close"] / first_build["First_Close"] - 1) * 100
).round(2)
first_build.index.name = "Ticker"
first_build = first_build.reset_index()
first_build["Company"] = first_build["Ticker"].map({
    "AMD": "AMD", "NVDA": "NVIDIA", "INTC": "Intel",
    "TSM": "TSMC", "AVGO": "Broadcom", "SPY": "SPY (S&P 500 Benchmark)"
})
first_build[["Ticker", "Company", "Cumulative_Return_Pct"]].sort_values("Ticker")
```

## Data Preparation

The raw yfinance data uses multi-level columns for price fields and ticker
symbols. The dataset is reshaped into long format so that each row represents
one ticker on one trading date. This structure makes the data easier to
analyze and visualize.

```python
clean_data = (
    data
    .stack(level="Ticker", future_stack=True)
    .reset_index()
)

clean_data.head(10)
```

```python
company_names = {
    "AMD": "AMD",
    "NVDA": "NVIDIA",
    "INTC": "Intel",
    "TSM": "TSMC",
    "AVGO": "Broadcom",
    "SPY": "SPY (S&P 500 Benchmark)"
}

clean_data["Company"] = clean_data["Ticker"].map(company_names)

clean_data.head(10)
```

```python
print("Dataset shape:", clean_data.shape)
print()

print("Date range:")
print("First date:", clean_data["Date"].min())
print("Last date:", clean_data["Date"].max())
print()

print("Rows by company:")
print(clean_data["Company"].value_counts())

print()
print("Missing values:")
print(clean_data.isnull().sum())
```

```python
clean_data = clean_data.sort_values(
    by=["Ticker", "Date"]
).reset_index(drop=True)
```

```python
clean_data["Daily_Return"] = (
    clean_data
    .groupby("Ticker")["Close"]
    .pct_change()
)
```

```python
clean_data["Daily_Return_Pct"] = clean_data["Daily_Return"] * 100
```

```python
clean_data[
    ["Date", "Ticker", "Company", "Close", "Daily_Return_Pct"]
].head(12)
```

## Cumulative Return

Cumulative return measures the total percentage change over the analysis period. It allows the historical performance of AMD, NVIDIA, Intel, TSMC, Broadcom, and the SPY benchmark to be compared from the same starting point. Historical performance is descriptive and does not guarantee future returns.


```python
clean_data["Cumulative_Return"] = (
    clean_data
    .groupby("Ticker")["Daily_Return"]
    .transform(lambda x: (1 + x.fillna(0)).cumprod() - 1)
)

clean_data["Cumulative_Return_Pct"] = (
    clean_data["Cumulative_Return"] * 100
)
```

```python
clean_data[
    [
        "Date",
        "Ticker",
        "Company",
        "Close",
        "Daily_Return_Pct",
        "Cumulative_Return_Pct"
    ]
].head(12)
```

```python
final_returns = (
    clean_data
    .groupby("Ticker")
    .tail(1)[
        ["Company", "Ticker", "Close", "Cumulative_Return_Pct"]
    ]
)

final_returns
```

## Annualized Volatility

Annualized volatility measures how much daily returns varied over the
historical period. Higher volatility indicates greater variability in
returns, while lower volatility indicates more stable returns.

Annualized volatility is calculated using the standard deviation of daily
returns multiplied by the square root of 252, representing approximately
252 trading days per year.

```python
annualized_volatility = (
    clean_data
    .groupby(["Ticker", "Company"])["Daily_Return"]
    .std()
    * np.sqrt(252)
    * 100
)

annualized_volatility = (
    annualized_volatility
    .reset_index(name="Annualized_Volatility_Pct")
)

annualized_volatility
```

```python
clean_data["Running_Peak"] = (
    clean_data
    .groupby("Ticker")["Close"]
    .cummax()
)
```

```python
clean_data["Drawdown"] = (
    clean_data["Close"] / clean_data["Running_Peak"]
) - 1

clean_data["Drawdown_Pct"] = (
    clean_data["Drawdown"] * 100
)
```

```python
max_drawdown = (
    clean_data
    .groupby(["Ticker", "Company"])["Drawdown_Pct"]
    .min()
    .reset_index(name="Max_Drawdown_Pct")
)

max_drawdown
```

```python
# Prepare cumulative return results
return_summary = final_returns[
    ["Ticker", "Company", "Cumulative_Return_Pct"]
].copy()

# Add annualized volatility
summary_table = return_summary.merge(
    annualized_volatility,
    on=["Ticker", "Company"]
)

# Add maximum drawdown
summary_table = summary_table.merge(
    max_drawdown,
    on=["Ticker", "Company"]
)

# Round values for readability
summary_table[
    [
        "Cumulative_Return_Pct",
        "Annualized_Volatility_Pct",
        "Max_Drawdown_Pct"
    ]
] = summary_table[
    [
        "Cumulative_Return_Pct",
        "Annualized_Volatility_Pct",
        "Max_Drawdown_Pct"
    ]
].round(2)

summary_table
```
